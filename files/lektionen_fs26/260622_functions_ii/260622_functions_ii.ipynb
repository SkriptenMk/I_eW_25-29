```python
#| code-fold: true
#| code-summary: Model answer
def does_the_student_pass(grades: list[float]) -> bool:
    """Determine whether a student qualifies for definitive admission or promotion.

    This function implements specific academic regulations based on a grading 
    scale where 4.0 is the passing mark. A student passes if both of the 
    following conditions are met:
    1. Twice the sum of all grade deviations below 4.0 is not greater than 
       the sum of all grade deviations above 4.0.
    2. No more than three grades are below 4.0.

    Args:
        grades (list[float]): A list of grades achieved in the promotional 
            subjects for the semester.

    Returns:
        bool: True if the promotion conditions are fulfilled, False otherwise.
    """
    sub_fours = [g - 4 for g in grades if g < 4]
    sup_fours = [g - 4 for g in grades if g >= 4]
    aggregated_sub_fours = abs(2 * sum(sub_fours))
    aggregated_sup_fours = sum(sup_fours)
    compensation = aggregated_sub_fours <= aggregated_sup_fours
    number_sub_fours = len(sub_fours)
    return compensation and (number_sub_fours <= 3)

# test cases
progress_secured = [5, 4.5, 4.5, 3.5, 5, 4, 4, 4, 4.5, 5, 5, 4]
failed_a = [5, 4.5, 4.5, 3.5, 4, 4, 4, 3, 4.5, 4, 4, 4]
failed_b = [5, 5, 4.5, 3.5, 4, 3.5, 4, 3, 4.5, 3.5, 5, 5.5]

print(f'The student progresses: {does_the_student_pass(progress_secured)}')
print(f'The student progresses a): {does_the_student_pass(failed_a)}')
print(f'The student progresses b): {does_the_student_pass(failed_b)}')
```
