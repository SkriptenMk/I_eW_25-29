# Inheritance

The use of auxiliary functions and the implementation of classes follow the
DRY (Do not Repeat Yourself) principle. But we can take the DRY
principle even further. We can use base classes blueprints for other
classes. We bundle common features of classes and create *child* classes
for specific new or extended features. This concept is known as
**inheritance**.

The following code cell shows an example based on the classes `Triangle`
and `Circle` from the previous section, which share the common base class `Shape`.

```python
import math
```

```python
class Shape:
    """Base class for all geometric shapes."""

    def get_area(self):
        """Calculates the area. Must be implemented by subclasses."""
        raise NotImplementedError("Subclasses must implement get_area().")

    def get_perimeter(self):
        raise NotImplementedError("Subclasses must implement get_perimeter().")

    def get_compactness(self):
        """Isoperimetric quotient: 1.0 for a circle, less for all others.

        Fully implemented here — works for ANY shape that knows its
        area and perimeter.
        """
        A = self.get_area()
        U = self.get_perimeter()
        return (4 * math.pi * A) / (U ** 2)
    
class Triangle(Shape):
    pass

class Circle(Shape):
    pass
```

The isoperimetric quotient is a measurement for the compactness of a
geometric figure[@doCarmo_Differentialgeometrie_98, pp. 26].
